```python
#Imputamos con media/mediana por grupos
df['peso_imputado_xsexo'] = df['peso'].fillna(
    df.groupby('sexo')['peso'].transform('mean')
)
df[['sexo', 'peso', 'peso_imputado_xsexo']].head(10)
```

```python
# Opción 3
# Imputamos mediante modelos predictivos

# Variables que usamos para predecir
variables = ['sexo', 'edad_num', 'talla_imputada_xsexo', 'peso']

df_modelo = df[variables].copy()

# Convertir sexo a numérica
df_modelo = pd.get_dummies(df_modelo, columns=['sexo'], drop_first=True)
```

```python
#Separar datos conocidos vs faltantes
df_train = df_modelo[df_modelo['peso'].notna()]
df_pred = df_modelo[
    (df_modelo['peso'].isna()) &
    (df_modelo.drop('peso', axis=1).notna().all(axis=1))
]

#Definir variables
X_train = df_train.drop('peso', axis=1)
y_train = df_train['peso']

X_pred = df_pred.drop('peso', axis=1)

#Entrenar modelo
modelo = LinearRegression()
modelo.fit(X_train, y_train)

#Predecir valores faltantes
peso_predicho = modelo.predict(X_pred)

#Imputar en el dataset original
df['peso_imputado_pred'] = df['peso']  # copiamos original

df.loc[df_pred.index, 'peso_imputado_pred'] = peso_predicho

#Verificamos
df[['peso', 'peso_imputado_pred', 'edad']].head(10)
```

### Imputación de datos: consideraciones importantes

La imputación consiste en reemplazar valores faltantes por valores estimados. Si bien es una práctica común, no siempre es adecuada y debe aplicarse con criterio.

En general, es aceptado imputar variables explicativas (features), como edad, talla o variables de laboratorio, especialmente cuando el porcentaje de datos faltantes es bajo y el método de imputación es apropiado.

Sin embargo, se debe ser especialmente cauteloso en los siguientes casos:

- Cuando la variable con datos faltantes es la variable respuesta (target). Imputarla puede introducir sesgos, ya que se estarían “inventando” resultados y el modelo podría aprender patrones artificiales.
- Cuando el porcentaje de datos faltantes es alto (mayor al 40%). En estos casos, los valores imputados pueden dominar el análisis y reducir la representatividad de los datos reales.
- Cuando los datos faltantes no son aleatorios. Por ejemplo, si ciertos pacientes no tienen mediciones por una razón clínica específica, imputar esos valores puede ocultar información relevante.
- En análisis inferenciales, donde la imputación puede afectar estimaciones, intervalos de confianza y valores p.
- En variables clínicas críticas, como diagnósticos o eventos, donde la imputación puede no ser apropiada.

Como regla general, es preferible imputar variables predictoras con métodos adecuados y evitar imputar la variable respuesta, salvo en contextos muy específicos y justificados.

En todos los casos, la imputación debe considerarse una decisión estadística que puede impactar en los resultados del análisis.

### **Generamos y transformamos variables**

```python
# Calculamos IMC

df['imc'] = df['peso'] / (df['talla'] ** 2)

df['imc'].describe()
```

```python
#Categorizamos el IMC

df['categoria_imc'] = pd.cut(
    df['imc'],
    bins=[0, 18.5, 25, 30, float('inf')],
    labels=['Bajo peso', 'Normopeso', 'Preobesidad', 'Obesidad'],
    right=False
)
```

```python
#Tabla de frecuencias

df['categoria_imc'].value_counts()
```

```python
#Tabla de frecuencias para presentar

tabla = df['categoria_imc'].value_counts().to_frame(name='Frecuencia absoluta')

tabla['Frecuencia relativa'] = df['categoria_imc'].value_counts(normalize=True)

tabla
```

### *Algunas pruebas estadísticas*


```python
# Test de normalidad

from scipy.stats import shapiro

stat, p = shapiro(df['imc'].dropna())

print('Estadístico:', stat)
print('p-valor:', p)
```

```python
sns.histplot(df['imc'], kde=True)
```

```python
# Test t, imc por sexo

#Vemos medias por sexo

df.groupby('sexo')['imc'].mean()
```

```python
from scipy.stats import ttest_ind

grupo_m = df[df['sexo'] == 'Masculino']['imc'].dropna()
grupo_f = df[df['sexo'] == 'Femenino']['imc'].dropna()
t_stat, p_valor = ttest_ind(grupo_m, grupo_f, equal_var=False)
print("t =", t_stat)
print("p-valor =", p_valor)
```

```python
# Test Mann-Whitney U

# Test
stat, p_valor = stats.mannwhitneyu(grupo_m, grupo_f, alternative='two-sided')

print(f"Estadístico U: {stat}")
print(f"p-valor: {p_valor}")
```

```python
# Test chi2, categorías de imc por sexo

from scipy.stats import chi2_contingency

# Tabla de contingencia
tabla = pd.crosstab(df['sexo'], df['categoria_imc'])
print("Tabla de contingencia:")
print(tabla)

# Test Chi-cuadrado
chi2, p_valor, gl, esperados = chi2_contingency(tabla)

print(f"\nEstadístico Chi2: {chi2:.4f}")
print(f"p-valor: {p_valor:.4f}")
print(f"Grados de libertad: {gl}")
```

```python
# Verificar valores esperados

esperados_df = pd.DataFrame(esperados, 
                             index=tabla.index, 
                             columns=tabla.columns)
print("\nFrecuencias esperadas:")
print(esperados_df)
print(f"\n¿Todos los esperados >= 5? {(esperados_df >= 5).all().all()}")
```
